```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
%matplotlib inline
```

```python
df = pd.read_csv('data1/bike-sharing-demand/train.csv')
```

```python
df.info()
```

```python
dropped_df = df.drop(['datetime','casual','registered'], axis=1)
dropped_df
```

```python
X = dropped_df.drop('count',axis=1).values
y = dropped_df['count']
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=50)
scaler = StandardScaler()
scaler.fit(X_train)
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
lr = LinearRegression()
lr.fit(X_train,y_train)
y_pred = lr.predict(X_test)
```

```python
mse = mean_squared_error(y_test,y_pred)
rmse = np.sqrt(mse)

rmse, mse
```

```python
r2_score(y_test, y_pred)
```

```python
pd.Series(data = np.abs(np.round(lr.coef_,1)), index=dropped_df.drop('count',axis=1).columns).sort_values(ascending=False)
```

```python
X = dropped_df.drop(['count', 'workingday', 'windspeed', 'temp'],axis=1).values
y = dropped_df['count']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=50)
scaler = StandardScaler()
scaler.fit(X_train)
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
lr = LinearRegression()
lr.fit(X_train,y_train)
y_pred = lr.predict(X_test)
```

```python
mse = mean_squared_error(y_test,y_pred)
rmse = np.sqrt(mse)

rmse, mse
```

```python
r2_score(y_test, y_pred)
```

```python
RMSE = np.sqrt(neg_mse_scores * (-1))
np.mean(RMSE), RMSE
```

현재까지 랜덤포레스트 단독 모델이 가장 낫다

| 모델         | 평가 지표 | 단독 예측 (Individual Prediction) | 교차 검증 (Cross-Validation) |
| :----------- | :-------- | :-------------------------------- | :--------------------------- |
| **리니어 모델** | RMSE      | 155.08                            | 163.01                       |
|              | R2        | 0.2619                            | -0.0539                      |
| **랜덤 포레스트** | RMSE      | 147.79                            | 166.08                       |
|              | R2        | 0.3296                            | -0.0975                      |

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures
```

```python
results = []
for degree in range(1,7):
    model_poly = Pipeline([
        ('poly',PolynomialFeatures(degree=degree, include_bias=False)),
        ('linear',LinearRegression())
    ])
    model_poly.fit(X_train, y_train)
    pred_poly = model_poly.predict(X_test)
    mse = mean_squared_error(y_test,pred_poly)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, pred_poly)
    results.append({
        'degree' : degree,
        'MSE' : mse,
        'RMSE' : rmse,
        'R2' : r2
    })
pd.DataFrame(results)
```

다항회귀모델(선형)에서는 4차가 가장 높다

```python
results = []
for degree in range(1,6):
    model_poly = Pipeline([
        ('poly',PolynomialFeatures(degree=degree, include_bias=False)),
        ('RF',RandomForestRegressor(random_state=22, max_depth=8))
    ])
    model_poly.fit(X_train, y_train)
    pred_poly = model_poly.predict(X_test)
    mse = mean_squared_error(y_test,pred_poly)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, pred_poly)
    results.append({
        'degree' : degree,
        'MSE' : mse,
        'RMSE' : rmse,
        'R2' : r2
    })
pd.DataFrame(results)
```

다항회귀모델(랜덤포레스트)에서는 단항이 가장 높다

결론적으로 랜덤포레스트 단독 모델일때 성능이 가장 높았다.
