```python
html = """
<html>
<head><title>HTML Sample</title>
</head>
<body>
    <h1>Hello CSS</h1>
    <div id="subject">선택자</div>
    <div class="contents">선택자를 어떻게 작성하느냐에 따라
      <span>다른 <b>요소가 반환</b></span> 됩니다.</div>
    <div>CSS 선택자는 다양한 곳에서 <b>활용</b>됩니다.</div>
</body>
</html>
"""
```

```python
html
```

```python
from bs4 import BeautifulSoup
soup = BeautifulSoup(html, 'html.parser')
```

```python
e1 = soup.select_one('h1')
e1.text
```

```python
e2 = soup.select('div')
for e in e2:
    print(e.text)
```

```python
e3 = soup.select('h1, span')
e3
```

```python
e4 = soup.select('div b') #후손
e4
```

```python
e5 = soup.select('div > b') #직계
e5
```

```python
e6 = soup.select('.contents')
e6
```

```python
e7 = soup.select('#subject')
e7
```

```python
e8 = soup.select('[id=subject]')
e8
```
